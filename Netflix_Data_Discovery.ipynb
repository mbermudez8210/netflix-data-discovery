# Netflix Data Discovery

This notebook explores Netflix's content library to identify trends in content production, genre popularity, and global distribution.

```
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: Set plotting style
sns.set(style='darkgrid')
```

```
# Load dataset
# Replace 'netflix_data.csv' with your actual CSV file name
netflix_df = pd.read_csv('netflix_data.csv')
netflix_df.head()
```

## Exploratory Data Analysis

```
# Check data info
netflix_df.info()
```

```
# Check for null values
netflix_df.isnull().sum()
```

```
# Top 10 genres
netflix_df['genre'].value_counts().head(10).plot(kind='bar', title='Top 10 Genres')
plt.show()
```
